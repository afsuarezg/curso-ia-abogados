## 1. Una regla, evaluada paso a paso

> **Regla 1.** Si la edad del votante es mayor o igual a 18 años, y si la nacionalidad del votante es "ciudadano", y si el registro del votante es "activo", entonces el votante es elegible para votar.

Tres premisas unidas por **y**. Cambie los hechos de las primeras líneas y vuelva a ejecutar.

````python
# Los hechos del caso
edad = 20
nacionalidad = "ciudadano"
registro = "pendiente"          # pruebe con "activo"

# Cada premisa se evalúa por separado y da verdadero (True) o falso (False)
premisas = [
    ("la edad es mayor o igual a 18 años", edad >= 18),
    ('la nacionalidad es "ciudadano"',     nacionalidad == "ciudadano"),
    ('el registro es "activo"',            registro == "activo"),
]

print("SUPUESTO DE HECHO")
for i, (texto, valor) in enumerate(premisas):
    print(f"  {'SI' if i == 0 else ' y'}  {texto}  ->  {valor}")

# La "y" exige que todas sean verdaderas
condicion = (edad >= 18) and (nacionalidad == "ciudadano") and (registro == "activo")
print(f"\n  La condición completa es {condicion}")

print("\nCONSECUENCIA")
if condicion:
    print("  => el votante es elegible para votar")
else:
    faltan = [t for t, v in premisas if not v]
    print("  => la regla no se aplica, porque no se cumple que " + " ni que ".join(faltan))
    print()
    print("  Ojo con lo que esto significa: la regla NO dice que la persona")
    print("  no pueda votar. Dice que esta regla no resuelve el caso. Para eso")
    print("  hacen falta otras reglas, y por eso el ejemplo de clase trae")
    print("  además una Regla 2 y una Regla 3.")
````

## 2. Y, O, NO: el conector cambia la norma

Con **y** hacen falta las dos premisas: son *requisitos copulativos*. Con **o** basta una: son *alternativos*. Y **no** sirve para escribir una excepción.

Cambiar una sola palabra en una norma cambia a quién se le aplica. Esta es la tabla completa.

````python
from itertools import product

A = "tiene 18 años o más"
B = "su registro está activo"
print(f"A = {A}")
print(f"B = {B}\n")

print(f"{'A':<8}{'B':<8}{'A y B':<8}{'A o B':<8}{'A y no B':<10}")
print("-" * 42)
for a, b in product([True, False], repeat=2):
    print(f"{str(a):<8}{str(b):<8}{str(a and b):<8}{str(a or b):<8}{str(a and not b):<10}")

combinaciones = list(product([True, False], repeat=2))
cuenta = lambda f: sum(1 for a, b in combinaciones if f(a, b))
print("\nDe las 4 combinaciones posibles:")
print(f"  «y»      la cumplen {cuenta(lambda a, b: a and b)}  -> la norma se aplica a muy pocos casos")
print(f"  «o»      la cumplen {cuenta(lambda a, b: a or b)}  -> se aplica a muchos más")
print(f"  «y no»   la cumplen {cuenta(lambda a, b: a and not b)}  -> igual de estrecha que «y», pero a otros casos")
````

## 3. Una cadena de reglas: el orden decide

Las reglas de culpabilidad del ejemplo de clase, puestas en fila. Se leen de arriba abajo y **gana la primera que se cumple**: las de abajo ni se miran.

A la lista original le agregamos al final una **regla general**, para ver en la celda siguiente qué pasa cuando se la sube de posición.

````python
# Los hechos del caso
pruebas = "insuficientes"       # "contundentes" | "insuficientes" | "indeterminadas"
coartada_solida = False
testigos_confiables = True

# Cada regla es (texto de la condición, función que la evalúa, consecuencia)
reglas = [
    ("las pruebas son contundentes",
     lambda: pruebas == "contundentes", "Culpable"),
    ("las pruebas son insuficientes y la coartada es sólida",
     lambda: pruebas == "insuficientes" and coartada_solida, "Inocente"),
    ("las pruebas son insuficientes, la coartada no es sólida y los testigos son confiables",
     lambda: pruebas == "insuficientes" and not coartada_solida and testigos_confiables, "Inocente"),
    ("las pruebas son insuficientes, la coartada no es sólida y los testigos no son confiables",
     lambda: pruebas == "insuficientes" and not coartada_solida and not testigos_confiables, "Culpable"),
    ("las pruebas no permiten determinar la culpabilidad",
     lambda: pruebas == "indeterminadas", "No se pudo determinar"),
    ("las pruebas son insuficientes",                      # <- la regla general que agregamos
     lambda: pruebas == "insuficientes", "Inocente"),
]

def aplicar(reglas):
    """Evalúa las reglas en orden y devuelve la consecuencia de la primera que se cumpla."""
    resultado, tapadas = None, 0
    for i, (texto, condicion, consecuencia) in enumerate(reglas, 1):
        if resultado is None and condicion():
            marca, resultado = "se aplica", consecuencia
        elif resultado is None:
            marca = "no se cumple"
        elif condicion():
            marca, tapadas = "tapada", tapadas + 1
        else:
            marca = "ni se mira"
        print(f"  [{marca:^12}] {i}. SI {texto}  ->  {consecuencia}")
    print(f"  Resultado: {resultado}")
    if tapadas:
        print(f"  «tapada» = también se cumple, pero quedó debajo de la que ganó y ya no se mira ({tapadas}).")
    return resultado

aplicar(reglas)
````

Ahora cambiamos **un solo hecho** —que los testigos no sean confiables— y comparamos los dos órdenes con los mismos hechos.

````python
testigos_confiables = False

print("ORDEN ORIGINAL (la regla general, de última):")
con_orden_original = aplicar(reglas)

reordenadas = [reglas[-1]] + reglas[:-1]
print("\nLA REGLA GENERAL SUBIDA AL PRIMER LUGAR:")
con_general_arriba = aplicar(reordenadas)

print()
print(f"  Mismos hechos, distinto orden: «{con_orden_original}» frente a «{con_general_arriba}».")
print("  Una regla general colocada arriba deja en letra muerta a todas las")
print("  especiales que vengan después. Es el mismo problema de redacción que")
print("  en derecho se resuelve con el principio de especialidad.")
````

## 4. Esto es todo lo que hace ELIZA

ELIZA (Weizenbaum, 1966) es una lista de reglas como las anteriores. Cada una tiene un **rango**, que es la prioridad que le puso el programador: se prueban de mayor a menor y gana la primera que encuentra su palabra clave.

Estas son cuatro reglas reales del ejemplo interactivo del curso, más la respuesta de relleno.

````python
import re

# (rango, palabras clave, respuesta) — ya ordenadas de mayor a menor rango
REGLAS_ELIZA = [
    (50, ["computadora", "computador", "máquina", "robot", "chatgpt"],
         "¿Te preocupan las computadoras?"),
    (10, ["abogado", "juez", "ley", "demanda", "contrato", "tribunal"],
         "¿Qué papel tiene la ley en lo que me cuentas?"),
    (3,  ["estoy"], "¿Cuánto tiempo llevas así?"),
    (2,  ["mi", "mis"], "¿Por qué dices que es tuyo?"),
]
GENERICA = "Cuénteme más."

def responder(frase):
    palabras = re.findall(r"\w+", frase.lower())
    print(f"Usted: {frase}")
    disparada = None
    for rango, claves, respuesta in REGLAS_ELIZA:
        encontrada = next((c for c in claves if c in palabras), None)
        if disparada is None and encontrada:
            marca, detalle = "se dispara", f": encontró la palabra «{encontrada}»"
            disparada = respuesta
        elif disparada is None:
            marca, detalle = "no se cumple", ": ninguna de sus palabras aparece"
        elif encontrada:
            marca, detalle = "tapada", f": «{encontrada}» también estaba, pero tiene menos rango"
        else:
            marca, detalle = "ni se mira", ""
        print(f"  [{marca:^13}] rango {rango:>2}{detalle}")
    if disparada is None:
        print(f"  [{'genérica':^13}] ninguna regla se cumplió: «observación libre de contenido»")
        disparada = GENERICA
    print(f"ELIZA: {disparada}\n")
    return disparada

responder("mi abogado dice que estoy preocupado por el contrato")
responder("me siento bastante raro hoy")
````

La regla solo mira si esas letras aparecen en la frase. ELIZA no sabe qué es una madre, ni qué es un contrato.

Es exactamente el mismo mecanismo del ejemplo de clase que decide el tipo de contrato buscando una palabra dentro del texto:

```python
if "arrendamiento" in texto_contrato.lower():
    tipo = "Contrato de arrendamiento"
```

## ¿Qué observar?

1. Cuando la condición es falsa, el condicional **no dice nada**. No dice lo contrario: dice que esa regla no resuelve el caso.
2. Cambiar una «y» por una «o» cambia a quién se le aplica la norma. Es la diferencia entre requisitos copulativos y alternativos, y se decide con una sola palabra.
3. En una cadena, **el orden decide**: una regla general puesta arriba deja sin efecto a todas las especiales que vengan después.
4. ELIZA es exactamente esto: una lista de reglas «si el texto contiene… entonces responder…», ordenadas por prioridad. No hay nada más adentro.
5. Un sistema de reglas es **predecible y auditable**: se pueden leer todas, una por una, y saber por qué respondió lo que respondió. Esa es su gran ventaja frente a un modelo que aprende — y, al mismo tiempo, su límite: nadie puede escribir a mano las reglas que harían falta para entender el lenguaje de verdad.
